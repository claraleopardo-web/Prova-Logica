Questão 1 — Classificação de Temperaturas (Condicionais + Loop)

```python
temperaturas = [18, 22, 30, 35, 28, 15, 40]
for temperatura in temperaturas:
  if (temperatura < 20):
    print("Frio")
  elif(temperatura >= 20 and temperatura <= 30):
    print("Agradável")
  elif(temperatura > 30):
    print("Quente")
```

```python
frio_count = 0
agradavel_count = 0
quente_count = 0

for temperatura in temperaturas:
  if (temperatura < 20):
    frio_count += 1
  elif(temperatura >= 20 and temperatura <= 30):
    agradavel_count += 1
  elif(temperatura > 30):
    quente_count += 1

print(f"Temperaturas Frias: {frio_count}")
print(f"Temperaturas Agradáveis: {agradavel_count}")
print(f"Temperaturas Quentes: {quente_count}")
```

Questão 2 — Sistema de Avaliação de Alunos

```python
notas_alunos = [4.5, 6.0, 7.8, 9.0, 5.5, 8.2]
for nota in notas_alunos:
  if(nota < 5):
    print("Reprovado")
  elif(nota >=5 and nota <= 7):
    print("Recuperação")
  elif(nota > 7):
    print("Aprovado")
```

```python
aprovados_count = 0
reprovados_count = 0

for nota in notas_alunos:
  if nota < 5:
    reprovados_count += 1
  elif nota > 7:
    aprovados_count += 1

total_alunos = len(notas_alunos)
percentual_reprovados = (reprovados_count / total_alunos) * 100 if total_alunos > 0 else 0

print(f"Quantidade de Alunos Aprovados: {aprovados_count}")
print(f"Percentual de Alunos Reprovados: {percentual_reprovados:.2f}%")
```

Questão 3 — Monitoramento de Consumo de Energia

```python
consumo = [120, 95, 200, 180, 75, 220, 160]
for item_consumo in consumo:
  if item_consumo < 100:
    print("Baixo")
  elif item_consumo >= 100 and item_consumo <= 180:
    print("Moderado")
  elif item_consumo > 180:
    print("Alto")
```

```python
print("Dias com consumo Alto:")
for i, item_consumo in enumerate(consumo):
  if item_consumo > 180:
    print(f"Dia {i+1}: {item_consumo} kWh")
```

```python
total_consumo = sum(consumo)
media_semanal = total_consumo / len(consumo)

print(f"Total de consumo da semana: {total_consumo} kWh")
print(f"Média semanal de consumo: {media_semanal:.2f} kWh")
```

```python
dias_alto_consumo = 0
for item_consumo in consumo:
  if item_consumo > 180:
    dias_alto_consumo += 1

if dias_alto_consumo > 2:
  print(f"ALERTA: Há {dias_alto_consumo} dias com consumo acima do normal!")
else:
  print(f"Não há alerta: Apenas {dias_alto_consumo} dias com consumo alto.")
```

Questão 4 — Simulação de Carrinho de Compras


```python
lista_precos = [50, 120, 30, 200, 80, 15]
for precos in lista_precos:
  if precos < 50:
    print("5% de desconto")
  elif precos >= 50 and precos <= 150:
    print("10% de desconto")
  elif precos > 150:
    print("15% de desconto")
```

```python
lista_precos = [50, 120, 30, 200, 80, 15]
print(f"Lista de preços originais: {lista_precos}")
```

```python
total_compra_original = sum(lista_precos)
total_compra_com_desconto = 0
economia_total = 0

for preco in lista_precos:
  if preco < 50:
    desconto = 0.05
  elif preco >= 50 and preco <= 150:
    desconto = 0.10
  else:
    desconto = 0.15

  preco_final = preco * (1 - desconto)
  total_compra_com_desconto += preco_final
  economia_total += (preco - preco_final)

print(f"Total da Compra com Desconto: {total_compra_com_desconto:.2f} ")
print(f"Economia Total Obtida: {economia_total:.2f} ")
```
